```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Model, load_model
from tensorflow.keras.preprocessing import image
import os
import glob
from tqdm import tqdm  # Progress bar
from collections import defaultdict

# Load the trained model
model = load_model("model_mobile.h5")

# Extract layers from the MobileNetV2 model
base_model = model.get_layer("mobilenetv2_1.00_224")  # Adjust the layer name if needed

early_feature_extractor = Model(inputs=base_model.input, outputs=base_model.get_layer("block_1_expand_relu").output)
middle_feature_extractor = Model(inputs=base_model.input, outputs=base_model.get_layer("block_6_expand_relu").output)
final_feature_extractor = Model(inputs=base_model.input, outputs=base_model.get_layer("out_relu").output)

# Define test image path
test_images_path = "./dataset/test"

# Define class labels
class_labels = ["freshapples", "freshbanana", "freshoranges", "rottenapples", "rottenbanana", "rottenoranges"]

# Create directories to store features
feature_dirs = {
    "early": "mobile_net_early_layer",
    "middle": "mobile_net_middle_layer",
    "final": "mobile_net_last_layer"
}

for dir_path in feature_dirs.values():
    os.makedirs(dir_path, exist_ok=True)

# Store selected image paths for each class
class_images = defaultdict(list)

# Collect images for each class
for class_name in class_labels:
    class_path = os.path.join(test_images_path, class_name)
    if os.path.exists(class_path):
        image_paths = glob.glob(os.path.join(class_path, "*.*"))  # Load image files only
        class_images[class_name] = image_paths[:100]  # Take first 100 images

# Extract and save features separately for each layer
for class_name, img_list in class_images.items():
    early_features_list = []
    middle_features_list = []
    final_features_list = []
    
    print(f" Extracting features for {class_name} ({len(img_list)} images)...")
    
    for img_path in tqdm(img_list, desc=f"Processing {class_name}"):
        try:
            # Load and preprocess image
            img = image.load_img(img_path, target_size=(224, 224))
            img_array = image.img_to_array(img)
            img_array = np.expand_dims(img_array, axis=0)  # Add batch dimension
            img_array = img_array / 255.0  # Normalize

            # Extract features
            early_features = early_feature_extractor.predict(img_array).flatten()
            middle_features = middle_feature_extractor.predict(img_array).flatten()
            final_features = final_feature_extractor.predict(img_array).flatten()

            # Store separately
            early_features_list.append(early_features)
            middle_features_list.append(middle_features)
            final_features_list.append(final_features)
        
        except Exception as e:
            print(f"Error processing {img_path}: {e}")

    # Convert to numpy arrays
    early_features_list = np.array(early_features_list)
    middle_features_list = np.array(middle_features_list)
    final_features_list = np.array(final_features_list)

    # Save features separately in corresponding directories
    np.savez_compressed(os.path.join(feature_dirs["early"], f"{class_name}_features.npz"), early_features_list)
    np.savez_compressed(os.path.join(feature_dirs["middle"], f"{class_name}_features.npz"), middle_features_list)
    np.savez_compressed(os.path.join(feature_dirs["final"], f"{class_name}_features.npz"), final_features_list)

    print(f" Features saved for {class_name}:")
    print(f"   - Early Layer: {feature_dirs['early']}/{class_name}_features.npz (Shape: {early_features_list.shape})")
    print(f"   - Middle Layer: {feature_dirs['middle']}/{class_name}_features.npz (Shape: {middle_features_list.shape})")
    print(f"   - Final Layer: {feature_dirs['final']}/{class_name}_features.npz (Shape: {final_features_list.shape})")

print(" Feature extraction complete for all classes!")
```

```python
import os
import numpy as np

# Define feature directories
feature_dirs = {
    "early": "mobile_net_early_layer",
    "middle": "mobile_net_middle_layer",
    "final": "mobile_net_last_layer"
}

# Expected number of images per class
EXPECTED_COUNT = 100

# Check feature files
for layer, dir_path in feature_dirs.items():
    print(f"\n🔍 Checking {layer} layer features...")
    
    for feature_file in os.listdir(dir_path):
        if feature_file.endswith(".npz"):
            file_path = os.path.join(dir_path, feature_file)

            # Load the feature file
            data = np.load(file_path)
            features = data[list(data.keys())[0]]  # Extract the stored array

            # Check shape
            if features.shape[0] != EXPECTED_COUNT:
                print(f"❌ MISMATCH: {feature_file} contains {features.shape[0]} instead of {EXPECTED_COUNT}")
            else:
                print(f"✅ {feature_file} is correct ({features.shape})")

print("\nFeature verification complete! ✅")
```

```python
# Define paths
early_layer_path = "mobile_net_early_layer"
middle_layer_path = "mobile_net_middle_layer"
last_layer_path = "mobile_net_last_layer"
output_path = "fused_features"

# Ensure output directory exists
os.makedirs(output_path, exist_ok=True)

# Get all class names from early layer folder
class_files = [f for f in os.listdir(early_layer_path) if f.endswith(".npz")]

# Process each class
for class_file in class_files:
    class_name = class_file.replace(".npz", "")  # Extract class name (e.g., freshapples)
    
    print(f"Processing {class_name}...")

    # Load features from all three layers
    early_features = np.load(os.path.join(early_layer_path, class_file))['arr_0']
    middle_features = np.load(os.path.join(middle_layer_path, class_file))['arr_0']
    last_features = np.load(os.path.join(last_layer_path, class_file))['arr_0']

    # Ensure all layers have the same number of samples
    num_samples = early_features.shape[0]
    if not (num_samples == middle_features.shape[0] == last_features.shape[0]):
        print(f" Mismatch in number of samples for {class_name}! Skipping...")
        continue

    # Apply PCA to reduce dimensionality
    pca = PCA(n_components=100)
    early_pca = pca.fit_transform(early_features)
    middle_pca = pca.fit_transform(middle_features)
    last_pca = pca.fit_transform(last_features)

    # Concatenate PCA-reduced features
    fused_features = np.concatenate([early_pca, middle_pca, last_pca], axis=1)

    # Save fused features with modified naming
    fused_filename = f"fused_{class_name}.npy"
    np.save(os.path.join(output_path, fused_filename), fused_features)

    print(f"✅ Saved: {fused_filename} (Shape: {fused_features.shape})")

print("🎉 Feature fusion complete!")     
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Model, load_model
from tensorflow.keras.preprocessing import image
import os
import glob
from tqdm import tqdm
from collections import defaultdict
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Load trained model
model = load_model("model_resnet50.h5")

# Get base ResNet50 model from trained model
base_model = model.get_layer("resnet50")

# Feature Extractor Models (Early, Middle, Final Layers)
early_feature_extractor = Model(inputs=base_model.input, outputs=base_model.get_layer("conv2_block1_out").output)
middle_feature_extractor = Model(inputs=base_model.input, outputs=base_model.get_layer("conv4_block1_out").output)
final_feature_extractor = Model(inputs=base_model.input, outputs=base_model.get_layer("conv5_block3_out").output)

# Define test image path
test_images_path = "./dataset/test"

# Define class labels
class_labels = ["freshapples", "freshbanana", "freshoranges", "rottenapples", "rottenbanana", "rottenoranges"]

# Store selected image paths for each class
class_images = defaultdict(list)

# Collect images for each class
for class_name in class_labels:
    class_path = os.path.join(test_images_path, class_name)
    if os.path.exists(class_path):
        image_paths = glob.glob(os.path.join(class_path, "*.*"))
        class_images[class_name] = image_paths[:100]  # Take first 100 images

# Extract and save features for each class
for class_name, img_list in class_images.items():
    all_features = []
    
    print(f" Extracting features for {class_name} ({len(img_list)} images)...")
    
    for img_path in tqdm(img_list, desc=f"Processing {class_name}"):
        try:
            # Load and preprocess image
            img = image.load_img(img_path, target_size=(224, 224))
            img_array = image.img_to_array(img)
            img_array = np.expand_dims(img_array, axis=0)  # Add batch dimension
            img_array = img_array / 255.0  # Normalize

            # Extract features
            early_features = early_feature_extractor.predict(img_array)
            middle_features = middle_feature_extractor.predict(img_array)
            final_features = final_feature_extractor.predict(img_array)

            # Apply Global Average Pooling (GAP)
            early_features = np.mean(early_features, axis=(1, 2))
            middle_features = np.mean(middle_features, axis=(1, 2))
            final_features = np.mean(final_features, axis=(1, 2))

            # Combine features
            combined_features = np.concatenate([early_features.flatten(), middle_features.flatten(), final_features.flatten()])
            all_features.append(combined_features)

        except Exception as e:
            print(f"⚠️ Error processing {img_path}: {e}")

    # Convert to numpy array
    all_features = np.array(all_features)

    # Normalize combined features
    scaler = StandardScaler()
    all_features = scaler.fit_transform(all_features)

    # Get the number of samples
    num_samples, num_features = all_features.shape

    # Set PCA components dynamically
    pca_early = PCA(n_components=min(100, num_samples))
    pca_middle = PCA(n_components=min(200, num_samples))
    pca_final = PCA(n_components=min(200, num_samples))

    early_pca = pca_early.fit_transform(all_features[:, :early_features.size])
    middle_pca = pca_middle.fit_transform(all_features[:, early_features.size:early_features.size + middle_features.size])
    final_pca = pca_final.fit_transform(all_features[:, early_features.size + middle_features.size:])

    fused_pca_features = np.concatenate([early_pca, middle_pca, final_pca], axis=1)

    # Save separately for Method 1 (Layer-wise PCA)
    fused_pca_filename = f"resnet_{class_name}_pca_fused(2).npy"
    np.save(fused_pca_filename, fused_pca_features)
    print(f" ✅ Features saved: {fused_pca_filename} (Shape: {fused_pca_features.shape})")

    # **Method 2: PCA After Fusion**
    pca_combined = PCA(n_components=min(500, num_samples))  # Dynamically adjust PCA components
    combined_pca_features = pca_combined.fit_transform(all_features)

    # Save separately for Method 2 (PCA After Fusion)
    combined_pca_filename = f"resnet_{class_name}_pca_combined(2).npy"
    np.save(combined_pca_filename, combined_pca_features)
    print(f" ✅ Features saved: {combined_pca_filename} (Shape: {combined_pca_features.shape})")

print("✅ Feature extraction complete for all classes!")

```
